```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

print("Imports successful")
```

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully")
```

```python
print(model)
```

```python
resume_text = """
Machine Learning Engineer with experience in Python, Pandas,
Scikit-learn, FastAPI and React.

Built machine learning applications for customer churn prediction
and used-car price estimation.

Experienced in data cleaning, feature engineering, model evaluation,
cross-validation and deploying ML models through REST APIs.
"""

job_description = """
We are looking for a Machine Learning Engineer with strong Python
and machine learning experience.

The candidate should have experience with Scikit-learn, feature
engineering, model evaluation, REST APIs and deploying machine
learning applications.

Experience with FastAPI, Docker and cloud deployment is preferred.
"""
```

```python
resume_embedding = model.encode(
    resume_text,
    convert_to_numpy=True
)

job_embedding = model.encode(
    job_description,
    convert_to_numpy=True
)

print("Resume embedding shape:", resume_embedding.shape)
print("Job embedding shape:", job_embedding.shape)
```

```python
similarity = cosine_similarity(
    [resume_embedding],
    [job_embedding]
)[0][0]

print(f"Semantic similarity: {similarity:.4f}")
print(f"Display score: {similarity * 100:.2f}%")
```

```python
skills = [
    "python",
    "pandas",
    "numpy",
    "scikit-learn",
    "machine learning",
    "feature engineering",
    "model evaluation",
    "cross-validation",
    "fastapi",
    "rest api",
    "react",
    "docker",
    "aws",
    "azure",
    "gcp",
    "cloud deployment",
    "sql",
    "git"
]
```

```python
def extract_skills(text, skill_list):
    text = text.lower()

    found_skills = []

    for skill in skill_list:
        if skill.lower() in text:
            found_skills.append(skill)

    return found_skills
```

```python
resume_skills = extract_skills(resume_text, skills)
job_skills = extract_skills(job_description, skills)

print("Resume skills:", resume_skills)
print("Job skills:", job_skills)
```

```python
matched_skills = sorted(
    set(resume_skills) & set(job_skills)
)

missing_skills = sorted(
    set(job_skills) - set(resume_skills)
)

print("Matched skills:", matched_skills)
print("Missing skills:", missing_skills)
```

```python
if len(job_skills) > 0:
    skill_match_score = (
        len(matched_skills) / len(job_skills)
    ) * 100
else:
    skill_match_score = 0

print(f"Skill match score: {skill_match_score:.2f}%")
```

```python
semantic_weight = 0.6
skill_weight = 0.4

overall_match_score = (
    similarity * 100 * semantic_weight
    + skill_match_score * skill_weight
)

print(f"Overall match score: {overall_match_score:.2f}%")
```

```python
if overall_match_score >= 80:
    match_level = "Strong Match"
elif overall_match_score >= 65:
    match_level = "Moderate Match"
else:
    match_level = "Weak Match"

print("Match level:", match_level)
```

```python
skill_aliases = {
    "python": ["python"],
    "pandas": ["pandas"],
    "numpy": ["numpy"],
    "scikit-learn": ["scikit-learn", "sklearn"],
    "machine learning": ["machine learning", "ml"],
    "feature engineering": ["feature engineering"],
    "model evaluation": ["model evaluation", "model validation"],
    "cross-validation": ["cross-validation", "cross validation"],
    "fastapi": ["fastapi"],
    "rest api": ["rest api", "restful api", "restful services"],
    "react": ["react", "react.js", "reactjs"],
    "docker": ["docker", "containerization"],
    "aws": ["aws", "amazon web services"],
    "azure": ["azure", "microsoft azure"],
    "gcp": ["gcp", "google cloud", "google cloud platform"],
    "cloud deployment": ["cloud deployment", "cloud computing"],
    "sql": ["sql", "postgresql", "mysql", "sqlite"],
    "git": ["git", "github"]
}
```

```python
def extract_normalized_skills(text, skill_aliases):
    text = text.lower()

    found_skills = []

    for canonical_skill, aliases in skill_aliases.items():
        for alias in aliases:
            if alias in text:
                found_skills.append(canonical_skill)
                break

    return sorted(set(found_skills))
```

```python
resume_skills = extract_normalized_skills(
    resume_text,
    skill_aliases
)

job_skills = extract_normalized_skills(
    job_description,
    skill_aliases
)

matched_skills = sorted(
    set(resume_skills) & set(job_skills)
)

missing_skills = sorted(
    set(job_skills) - set(resume_skills)
)

skill_match_score = (
    len(matched_skills) / len(job_skills) * 100
    if job_skills else 0
)

print("Resume skills:", resume_skills)
print("Job skills:", job_skills)
print("Matched skills:", matched_skills)
print("Missing skills:", missing_skills)
print(f"Skill match score: {skill_match_score:.2f}%")
```

```python
semantic_score = similarity * 100

overall_match_score = (
    semantic_score * 0.6
    + skill_match_score * 0.4
)

if overall_match_score >= 80:
    match_level = "Strong Match"
elif overall_match_score >= 65:
    match_level = "Moderate Match"
else:
    match_level = "Weak Match"

print(f"Semantic score: {semantic_score:.2f}%")
print(f"Skill match score: {skill_match_score:.2f}%")
print(f"Overall match score: {overall_match_score:.2f}%")
print("Match level:", match_level)
```

```python
def evaluate_match(resume_text, job_description):
    resume_embedding = model.encode(
        resume_text,
        convert_to_numpy=True
    )

    job_embedding = model.encode(
        job_description,
        convert_to_numpy=True
    )

    similarity = cosine_similarity(
        [resume_embedding],
        [job_embedding]
    )[0][0]

    semantic_score = similarity * 100

    resume_skills = extract_normalized_skills(
        resume_text,
        skill_aliases
    )

    job_skills = extract_normalized_skills(
        job_description,
        skill_aliases
    )

    matched_skills = sorted(
        set(resume_skills) & set(job_skills)
    )

    missing_skills = sorted(
        set(job_skills) - set(resume_skills)
    )

    skill_match_score = (
        len(matched_skills) / len(job_skills) * 100
        if job_skills else 0
    )

    overall_score = (
        semantic_score * 0.6
        + skill_match_score * 0.4
    )

    if overall_score >= 80:
        match_level = "Strong Match"
    elif overall_score >= 65:
        match_level = "Moderate Match"
    else:
        match_level = "Weak Match"

    return {
        "semantic_score": round(semantic_score, 2),
        "skill_match_score": round(skill_match_score, 2),
        "overall_score": round(overall_score, 2),
        "match_level": match_level,
        "matched_skills": matched_skills,
        "missing_skills": missing_skills
    }
```

```python
strong_resume = """
Machine Learning Engineer experienced in Python, Scikit-learn,
Pandas, NumPy, FastAPI, Docker, REST APIs, feature engineering,
model evaluation, cross-validation and cloud deployment on AWS.
"""

strong_job = """
We need a Machine Learning Engineer with Python, Scikit-learn,
feature engineering, model evaluation, FastAPI, REST API development,
Docker and AWS cloud deployment experience.
"""

evaluate_match(strong_resume, strong_job)
```

```python
weak_resume = """
English Literature lecturer with experience in teaching,
academic writing, curriculum development and literary criticism.
"""

weak_job = """
We need a Machine Learning Engineer experienced in Python,
Scikit-learn, Docker, FastAPI, SQL, AWS and machine learning.
"""

evaluate_match(weak_resume, weak_job)
```
